Tests for macro F1 score 

```python
# Build score dictionary: one entry per representation+preprocessing+model combination
all_scores = {}
for rep_prep, df in files.items():
    for model in df['model'].unique():
        key = f"{rep_prep} + {model}"
        all_scores[key] = df[df['model'] == model]['CV mF1'].tolist()

# Build matrix: rows = folds, columns = configurations
score_df = pd.DataFrame(all_scores)
print("Configurations:")
for col in score_df.columns:
    print(f"  {col}")
print(f"\nMatrix shape: {score_df.shape} (folds x configurations)")

#Friedman test 
print("FRIEDMAN TEST")
stat, p_mf1 = friedmanchisquare(*[score_df[col].values for col in score_df.columns])
print(f"Statistic: {stat:.4f}")
print(f"P-value:   {p_mf1:.6f}")
if p_mf1 < 0.05:
    print("Result: Significant difference found (p < 0.05)")
else:
    print("Result: No significant difference found (p >= 0.05)")

# Post-hoc Nemenyi 
if p_mf1 < 0.05:
    print("NEMENYI POST-HOC TEST (p-values)")
    nemenyi = posthoc_nemenyi_friedman(score_df)
    nemenyi.columns = score_df.columns
    nemenyi.index = score_df.columns

    # Significant pairs
    print("\nSignificant pairs (p < 0.05):")
    cols = score_df.columns.tolist()
    found_any = False
    for i in range(len(cols)):
        for j in range(i+1, len(cols)):
            p_val = nemenyi.iloc[i, j]
            if p_val < 0.05:
                print(f"  {cols[i]:45s} vs  {cols[j]:45s}  p = {p_val:.4f}")
                found_any = True
    if not found_any:
        print("  None found")

    # Save full Nemenyi matrix
    nemenyi.round(4).to_csv('../../data/output/nemenyi_mF1_results.csv')
    
    print("MEAN CV MACRO-F1 PER CONFIGURATION (ranked)")
    means = score_df.mean().sort_values(ascending=False)
    for config, mean_f1 in means.items():
        print(f"  {config:50s}  mean mF1 = {mean_f1:.4f}")
```

```python
# Reload nemenyi 
nemenyi = pd.read_csv('../../data/output/nemenyi_mF1_results.csv', index_col=0)

# Mean ranks across folds (Friedman ranks each configuration per fold)
mean_ranks = score_df.rank(ascending=False, axis=1).mean()

# Shorten names for readability in the plot
name_map = {
    'TF-IDF Lemma + Logistic Regression': 'TF-IDF Lemma + LR',
    'TF-IDF Lemma + Random Forest':       'TF-IDF Lemma + RF',
    'TF-IDF Lemma + SVM':                 'TF-IDF Lemma + SVM',
    'TF-IDF Stem + Logistic Regression':  'TF-IDF Stem + LR',
    'TF-IDF Stem + Random Forest':        'TF-IDF Stem + RF',
    'TF-IDF Stem + SVM':                  'TF-IDF Stem + SVM',
    'BoW Lemma + Logistic Regression':    'BoW Lemma + LR',
    'BoW Lemma + Random Forest':          'BoW Lemma + RF',
    'BoW Lemma + SVM':                    'BoW Lemma + SVM',
    'BoW Stem + Logistic Regression':     'BoW Stem + LR',
    'BoW Stem + Random Forest':           'BoW Stem + RF',
    'BoW Stem + SVM':                     'BoW Stem + SVM',
    'Word2Vec Lemma + Logistic Regression': 'W2V Lemma + LR',
    'Word2Vec Lemma + Random Forest':       'W2V Lemma + RF',
    'Word2Vec Lemma + SVM':                 'W2V Lemma + SVM',
    'Word2Vec Stem + Logistic Regression':  'W2V Stem + LR',
    'Word2Vec Stem + Random Forest':        'W2V Stem + RF',
    'Word2Vec Stem + SVM':                  'W2V Stem + SVM',
}

mean_ranks.index = [name_map.get(i, i) for i in mean_ranks.index]
nemenyi.index    = [name_map.get(i, i) for i in nemenyi.index]
nemenyi.columns  = [name_map.get(i, i) for i in nemenyi.columns]
```

Tests for balanced accuracy metric 

```python
# Build score dictionary: one entry per representation+preprocessing+model combination
all_scores = {}
for rep_prep, df in files.items():
    for model in df['model'].unique():
        key = f"{rep_prep} + {model}"
        all_scores[key] = df[df['model'] == model]['CV bACC'].tolist()

# Build matrix: rows = folds, columns = configurations
score_df = pd.DataFrame(all_scores)
print("Configurations:")
for col in score_df.columns:
    print(f"  {col}")
print(f"\nMatrix shape: {score_df.shape} (folds x configurations)")

# Friedman test 
print("FRIEDMAN TEST")
stat, p = friedmanchisquare(*[score_df[col].values for col in score_df.columns])
print(f"Statistic: {stat:.4f}")
print(f"P-value:   {p:.6f}")
if p < 0.05:
    print("Result: Significant difference found (p < 0.05)")
else:
    print("Result: No significant difference found (p >= 0.05)")

# Post-hoc Nemenyi
if p < 0.05:
    print("NEMENYI POST-HOC TEST (p-values)")
    nemenyi = posthoc_nemenyi_friedman(score_df)
    nemenyi.columns = score_df.columns
    nemenyi.index = score_df.columns

    # Significant pairs
    print("\nSignificant pairs (p < 0.05):")
    cols = score_df.columns.tolist()
    found_any = False
    for i in range(len(cols)):
        for j in range(i+1, len(cols)):
            p_val = nemenyi.iloc[i, j]
            if p_val < 0.05:
                print(f"  {cols[i]:45s} vs  {cols[j]:45s}  p = {p_val:.4f}")
                found_any = True
    if not found_any:
        print("  None found")

    # Save full Nemenyi matrix
    nemenyi.round(4).to_csv('../../data/output/nemenyi_bACC_results.csv')

    print("MEAN CV BALANCED ACCURACY PER CONFIGURATION (ranked)")
    means = score_df.mean().sort_values(ascending=False)
    for config, mean_bacc in means.items():
        print(f"  {config:50s}  mean bACC = {mean_bacc:.4f}")
```

```python
mf1  = pd.read_csv('../../data/output/nemenyi_mF1_results.csv',  index_col=0)
bacc = pd.read_csv('../../data/output/nemenyi_bACC_results.csv', index_col=0)

cols = mf1.columns.tolist()
alpha = 0.05

# For each pair, classify the significance under each metric 
rows = []
for i in range(len(cols)):
    for j in range(i+1, len(cols)):
        p_mf1  = mf1.iloc[i, j]
        p_bacc = bacc.iloc[i, j]
        sig_mf1  = p_mf1  < alpha
        sig_bacc = p_bacc < alpha

        if sig_mf1 and sig_bacc:
            agreement = 'Both significant'
        elif not sig_mf1 and not sig_bacc:
            agreement = 'Both non-significant'
        else:
            agreement = '*** DISAGREE ***'

        rows.append({
            'Config A': cols[i],
            'Config B': cols[j],
            'p (mF1)':  round(p_mf1,  4),
            'p (bACC)': round(p_bacc, 4),
            'Sig mF1':  sig_mf1,
            'Sig bACC': sig_bacc,
            'Agreement': agreement
        })

comparison_df = pd.DataFrame(rows)

print("AGREEMENT SUMMARY")

counts = comparison_df['Agreement'].value_counts()
for label, count in counts.items():
    print(f"  {label}: {count} pairs")

disagreements = comparison_df[comparison_df['Agreement'] == ' DISAGREE ']
print(f"\n{'='*60}")
print(f"DISAGREEING PAIRS ({len(disagreements)} total)")
if len(disagreements) == 0:
    print("  None: both metrics agree on all pairs.")
else:
    for _, row in disagreements.iterrows():
        print(f"\n  {row['Config A']}")
        print(f"  vs {row['Config B']}")
        print(f"  p(mF1)={row['p (mF1)']:.4f} (sig={row['Sig mF1']}),  "
              f"p(bACC)={row['p (bACC)']:.4f} (sig={row['Sig bACC']})")

# Show pairs significant under both metrics 
both_sig = comparison_df[comparison_df['Agreement'] == 'Both significant']
print(f"SIGNIFICANT UNDER BOTH METRICS ({len(both_sig)} pairs)")
for _, row in both_sig.iterrows():
    print(f"  {row['Config A']:45s} vs  {row['Config B']:45s} "
          f"p(mF1)={row['p (mF1)']:.4f}, p(bACC)={row['p (bACC)']:.4f}")

# Save full comparison
comparison_df.to_csv('../../data/output/nemenyi_metric_comparison.csv', index=False)
```

```python
configs = [
    (tfidf_lemma_lr,  'LR + TF-IDF + Lemma',  'blue',   0.698),
    (tfidf_stem_lr,   'LR + TF-IDF + Stem',   'green',  0.700),
    (tfidf_lemma_svm, 'SVM + TF-IDF + Lemma', 'red',    0.693),
    (bow_lemma_svm,   'SVM + BoW + Lemma',    'orange', 0.658),
]

fig, ax = plt.subplots(figsize=(7, 6))

for df_roc, label, color, reported_auc in configs:
    fpr, tpr, _ = roc_curve(df_roc['y_true'], df_roc['y_proba'])
    ax.plot(fpr, tpr, color=color, lw=2, label=f'{label} (AUC = {reported_auc:.3f})')

ax.plot([0, 1], [0, 1], color='grey', lw=1, linestyle='--', label='Baseline (AUC = 0.500)')
ax.set_xlim([0.0, 1.0])
ax.set_ylim([0.0, 1.05])
ax.set_xlabel('False Positive Rate (1 - Specificity)', fontsize=11)
ax.set_ylabel('True Positive Rate (Recall)', fontsize=11)
ax.set_title('ROC Curves', fontsize=12)
ax.legend(loc='lower right', fontsize=10)
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig(f'../../data/output/best_roc_curves_top4.png', dpi=300, bbox_inches='tight')
plt.show()
```
